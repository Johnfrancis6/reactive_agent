 Short and intentional.
 `logging.getLogger(name)` returns the same logger object every time you pass the same name — that's Python's logging registry. The `if not logger.handlers` guard prevents adding a duplicate handler if `get_logger` is called twice with the same name, which would print every log line twice.
 `logger.propagate = False` stops the log record from bubbling up to the root logger. Without this, if the root logger also has a handler configured, you get duplicate output again.
 The format — `asctime [LEVEL] name: message` — gives you timestamp, severity, which module logged it, and the message. Enough to debug without being noisy.
 **Usage everywhere else in the project:**
 ```python
 logger = get_logger(__name__)
 logger.info("Something happened: value=%s", some_value)
 ```
 The `%s` style — not f-strings. Python's logging is lazy: it only formats the string if the message actually gets emitted. With an f-string, you format it even if the log level would suppress it.

````python
import  logging
from logging  import Logger


def get_logger(name:str)-> Logger:
    logger = logging.getLogger(name)
    if not logger.handlers:
        handler = logging.StreamHandler()
        formatter =  logging.Formatter(
            "%(asctime)s [%(levelname)s] %(name)s: %(message)s"
        )
        handler.setFormatter(formatter)
        logger.addHandler(handler)
        logger.propagate = False
        logger.setLevel(logging.INFO)
    return logger
````
